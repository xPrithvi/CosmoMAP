# Extract Sample

```python
import numpy as np
import pandas as pd
```

# Load Files

```python
# Creating dataframe,
df = pd.read_csv("CosmoAstroSeed_IllustrisTNG_L25n256_EX.txt", sep="\\s+")

# Viewing dataframe,
df
```

# Creating Samples

```python
# Loading data files,
DMD_grids = np.load(r"Grids_Mcdm_IllustrisTNG_EX_256_z=0.0.npy", mmap_mode='r')
DMV_grids = np.load(r"Grids_Vcdm_IllustrisTNG_EX_256_z=0.0.npy", mmap_mode='r')
Ne_grids = np.load(r"Grids_ne_IllustrisTNG_EX_256_z=0.0.npy", mmap_mode='r')
T_grids = np.load(r"Grids_T_IllustrisTNG_EX_256_z=0.0.npy", mmap_mode='r')

# Simulation index,
j = 1
SIM_NAME = df["#Name"][j]

# Only considering first simulation,
DMD_grid, DMV_grid, Ne_grid, T_grid = DMD_grids[j], DMV_grids[j], Ne_grids[j], T_grids[j]

# Saving as .npy files,
np.save(f"DMD_grid({SIM_NAME}).npy", DMD_grid)
np.save(f"DMV_grid({SIM_NAME}).npy", DMV_grid)
np.save(f"Ne_grid({SIM_NAME}).npy", Ne_grid)
np.save(f"T_grid({SIM_NAME}).npy", T_grid)
```
