# Extract Sample

```python
import numpy as np
```

# Creating Samples

```python
# IMPORTANT - Specify redshifts,
z = [0.0, 0.5, 1.0, 1.5, 2.0]

for redshift in z:

    # Loading data files,
    DMD_grids = np.load(f"IllustrisTNG_CV/Grids_Mcdm_IllustrisTNG_CV_256_z={redshift}.npy", mmap_mode='r')
    DMV_grids = np.load(f"IllustrisTNG_CV/Grids_Vcdm_IllustrisTNG_CV_256_z={redshift}.npy", mmap_mode='r')
    Ne_grids = np.load(f"IllustrisTNG_CV/Grids_ne_IllustrisTNG_CV_256_z={redshift}.npy", mmap_mode='r')
    T_grids = np.load(f"IllustrisTNG_CV/Grids_T_IllustrisTNG_CV_256_z={redshift}.npy", mmap_mode='r')

    # Only considering first simulation,
    DMD_grid, DMV_grid, Ne_grid, T_grid = DMD_grids[0], DMV_grids[0], Ne_grids[0], T_grids[0]

    # Saving as .npy files,
    np.save(f"DMD_grid_z{redshift}.npy", DMD_grid)
    np.save(f"DMV_grid_z{redshift}.npy", DMV_grid)
    np.save(f"Ne_grid_z{redshift}.npy", Ne_grid)
    np.save(f"T_grid_z{redshift}.npy", T_grid)
```
